## Come avviare le versioni

Serve Python 3 con Tkinter. Le versioni si trovano nella cartella `codice/`, insieme a `enigmi.json`, che ogni versione legge dalla propria cartella. La cella seguente definisce la funzione `avvia`, che apre il gioco in una finestra separata.

```python
import subprocess
import sys
from pathlib import Path


def avvia(file):
    """Avvia una versione del gioco in una finestra separata."""
    percorso = Path(file).resolve()
    subprocess.Popen([sys.executable, str(percorso)], cwd=percorso.parent)
```
